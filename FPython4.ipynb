```python
# 1 Su primera clase
# Complete la definición para que la clase guarde el nombre y la edad de una persona.
class Persona:
 def __init__(self, nombre, edad):
  self.nombre = nombre
  self.edad = edad
p = Persona("Ana", 21)
print(p.nombre, p.edad)
```

```python
# 2 Crear un objeto
# Complete la línea que crea el objeto usando la clase ya definida.
p = Persona("Luis", 30)
print(p.nombre)
```

```python
# 3 Leer un atributo
# Complete las dos instrucciones para mostrar cada dato guardado en el objeto.
p = Persona("Sara", 25)
print(p.nombre)
print(p.edad)
```

```python
# 4 Modificar un atributo
# Cambie la edad guardada en el objeto y compruebe que el cambio quedó registrado.
p = Persona("Luis", 30)
print(p.edad)
p. = 31
print(p.edad)
```
