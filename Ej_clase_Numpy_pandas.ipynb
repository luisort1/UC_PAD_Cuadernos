```python
# Indexación, rebanado y máscaras
precios = np.array([24000, 28000, 22000, 30000, 8000])

print(precios[0])       # 24000        primer elemento
print(precios[-1] )      # 8000         último elemento
print(precios[1:4] )     # [28000 22000 30000]   el final se excluye

mascara = precios > 20000     # [True True True True False]
precios[mascara]              # [24000 28000 22000 30000]
precios[mascara].size         # 4

categoria = np.where(precios > 25000, "alto", "normal")
# ['normal' 'alto' 'normal' 'alto' 'normal']
# Para combinar condiciones se usan & y | , nunca and ni or,
# y cada comparación va entre paréntesis.
# Es exactamente la misma regla que se aplicará después sobre un DataFrame.
categoria.dtype
```

```python
# máscaras (filtros)
mascara_2 = (precios > 5000) & (precios < 26000) # &: para Y y | para O
print(f"Resultado de la mascara #2: {mascara_2}")
# máscaras (filtros)
mascara_3 = (precios > 5000) | (precios < 26000) # &: para Y y | para O
print(f"Resultado de precios  : {precios}")
print(f"Resultado de la mascara #3: {mascara_3}")
```
